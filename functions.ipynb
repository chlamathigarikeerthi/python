```python
#Function + loop
def multiplication_table(number):
    for i in range(1, 11):
        print(number, "x", i, "=", number * i)
number = int(input("Enter nummber: "))

multiplication_table(number)        
```

```python
#Largest of three numbers
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
a = int(input("Enter a: "))    
b = int(input("Enter b: "))    
c = int(input("Enter c: "))   
print("Largest =", largest(a, b, c)) 
```

```python
#Largest of two numbers
def largest(a, b):
    if a >= b:
        return a
    else:
        return b
a = int(input("Enter a: "))    
b = int(input("Enter b: "))       
print("Largest =", largest(a, b))     
```

```python
#Grade 
def find_grade(marks):
    if marks >= 90:
        return "Grade A"
    elif marks >= 80:
       return "Grade B"    
    elif marks >= 70:
       return "Grade C"
    elif marks >= 60:
       return "Grade D"
    elif marks >= 40:
       return "Grade E"
    else:
       return "fail"

marks = int(input("Enter marks: "))
print("Grade =", find_grade(marks))
```
