```python
#list in python
#list is an ordered and changeable collection that can store multiple values 
marks = [80, 90, 75, 85]
print(marks)
```

```python
#accessing elements in a list
marks = [80, 90, 75, 85]
print(marks[0]) 
print(marks[1]) 
print(marks[2]) 
```

```python
#change elements in a list
marks = [80, 90, 75,]
marks[1] = 95
print(marks)
```

```python
#change elements in a list
marks = [79,64,92,74,96,77,94]
marks[1] = 95
marks[3] = 100
marks[5] = 88
print(marks)
```

```python
#add elements to a list
marks = [80, 90, 75,]
marks.append(85)
print(marks)
```

```python
#adding elements to a list
marks = [86, 92, 78, 95]
marks.append(88)
print(marks)
```

```python
#adding elements to a list
marks = [80, 90, 75, 53, 85]
marks.append(95)
print(marks)
```

```python
#adding elements to a list
marks = [80, 90, 75, 53, 85]
marks.append(95)
print(marks)
```

```python
#adding elements to a list
marks = [40, 80, 55, 63, 85]
marks.append(95)
print(marks)
```

```python
#remove elements from a list
marks = [80, 90, 75]
marks.remove(90)
print(marks)
```

```python
numbers = [10, 20, 30]
numbers.insert(1, 15)
print(numbers)
```

```python
numbers = [10, 20, 30]
numbers.insert(2, 25)
numbers.insert(3, 35)
print(numbers)
```

```python
a = [1, 2, 3,]
b = [4, 5, 6]
a.extend(b)
print(a)
```

```python
numbers = [20, 20, 30, 20]
numbers.remove(20)
print(numbers)
```

```python
numbers = [10, 20, 30]
numbers.clear()
print(numbers)
```

```python
#index numbers
numbers = [10, 20, 30, 40]
print(numbers.index(30))
```

```python
#count numbers
numbers = [10, 20, 20, 30, 20]
print(numbers.count(20))
```

```python
#sort numbers
numbers = [40, 10, 30, 20]
numbers.sort()
print(numbers)
```

```python
#sort
numbers = [40, 10, 30, 20]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)
```

```python
a = [1, 2, 3]
b = a.copy()
print(b)
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4]) 
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])
```
